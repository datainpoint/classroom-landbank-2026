## indexing 的語法

`ndarray` 採用兩個方向的索引機制：

1. 由左至右：「從 0 開始」的索引機制。
2. 由右至左：「從 -1 開始」的索引機制。

````python
print("From start to stop:")
print(prime_array[0])
print(prime_array[1])
print(prime_array[2])
print(prime_array[3])
print(prime_array[prime_array.size - 1])
````

````python
print("From stop to start:")
print(prime_array[-1])
print(prime_array[-2])
print(prime_array[-3])
print(prime_array[-4])
print(prime_array[-prime_array.size])
````

## slicing `[start:stop:step]`

除了可以取出特定位置的單個資料值，`ndarray` 也支援擷取特定片段，藉此獲得一個較短長度 `ndarray` 的語法。

- `start` 起始位置（包含）。
- `stop` 終止位置（排除）。
- `step` 間隔。

````python
print(prime_array[::])   # default
print(prime_array[::2])  # step=2
print(prime_array[:3])   # stop=5, exclusive
print(prime_array[3:])   # start=5, inclusive
print(prime_array[::-1]) # step=-1, reverse
````

## 能夠以 indexing 更新

````python
print(prime_array)   # before update
prime_array[-1] = 13 # update
print(prime_array)   # after update
````

- 同質性資料結構類別。
- 支援元素操作（Elementwise）運算。
- 支援特殊的 indexing 語法。

## indexing 二維以上的 `ndarray` 可以用 `ndarray` 便捷的語法 `[i, j, k, ...]`

````python
matrix = np.array([[5, 5],
                   [6, 6],
                   [55, 66]])
print(matrix)
print(matrix[2, 1]) # 66 locates at [2, 1]
````

````python
print(matrix[:, 1])
print(matrix[:, [1]]) # keep dimension
````

## 同質性資料結構類別

````python
heterogeneous_list = [False, True, 5566, 55.66, 'Luke Skywalker']
for element in heterogeneous_list:
    print(type(element))
````

````python
homogeneous_array = np.array(heterogeneous_list)
for element in homogeneous_array:
    print(type(element))
````

## 支援元素操作（Elementwise）運算

````python
# list does not support elementwise
try:
    prime_list**2
except TypeError as error_message:
    print(error_message)
````

````python
# ndarray supports elementwise
prime_array = np.array(prime_list)
prime_array**2
````

## 支援特殊的 indexing 語法

- Fancy indexing
- Boolean indexing

## 什麼是 Fancy indexing

對應 `ndarray` 時中括號允許傳入 `list`，藉此可以更有彈性地取出 `ndarray` 中的元素。

````python
try:
    prime_list[[0, 1, 4]]
except TypeError as error_message:
    print(error_message)
````

````python
print(prime_array[[0, 1, 4]])
````

## 什麼是 Boolean indexing

對應 `ndarray` 時中括號允許傳入由 `bool` 組成的相同長度 `list` 或 `ndarray`，藉此可以更有彈性地取出 `ndarray` 中的元素。

````python
prime_list = [2, 3, 5, 7, 11]
try:
    prime_list[[False, True, True, True, True]]
except TypeError as error_message:
    print(error_message)
````

````python
prime_array = np.array([2, 3, 5, 7, 11])
print(prime_array)
print(prime_array[[False, True, True, True, True]])
print(prime_array % 2 == 1)
print(prime_array[prime_array % 2 == 1])
````

- 運用資料列數思考條件的結合要「縮減」還是「擴增」，「縮減」為交集、「擴增」為聯集。

## 實作二步驟：使用 Python 的做法

- 首先將 `listing_dates.csv` 上傳至 Google Colab 的工作目錄。
- 載入 `pandas` 模組並且取別名為 `pd`
- 使用 `pd.read_csv()` 函數載入為 `DataFrame`
- 使用 `Series.notna()` 方法辨識上市上櫃日期是否「非未定義值」。
- 使用 `datetime.today()` 函數取得今天日期，並轉換成 `Series`。
- 使用 `pd.to_datetime()` 函數將上市上櫃日期轉換成可計算的日期型別。
- 將今天日期與上市上櫃日期相減，可以得到掛牌天數，判斷掛牌天數是否超過 2 年（`365 * 2`）。
- 以 `&` 運算符交集「有上市上櫃日期」與「掛牌天數超過 2 年」這兩個條件。
- 利用布林索引剔除「沒有上市上櫃日期」以及「掛牌不到兩年的公司」。

```python
from datetime import date
import pandas as pd

file_path = "listing_dates.csv"
listing_dates_csv = pd.read_csv(file_path)
today = date.today()
today_ser = pd.to_datetime(today)
listing_date_ser = pd.to_datetime(listing_dates_csv["listing_date"])
listing_days = today_ser - listing_date_ser
listing_days = listing_days.dt.days
listing_date_notna = listing_dates_csv["listing_date"].notna()
days_of_two_years = 365*2
is_listing_days_over_two_years = listing_days > days_of_two_years
conditions_intersected = listing_date_notna & is_listing_days_over_two_years
listing_dates_csv[conditions_intersected]
```
